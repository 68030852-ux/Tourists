```python
import pandas as pd

df = pd.read_csv("c:\\Users\\supin\\Downloads\\tourists_all.csv")

df["Number"] = df["Number"].str.strip()
df["Number"] = df["Number"].str.replace(",", "")
df["Number"] = df["Number"].replace("-", pd.NA)

# แปลงเป็นตัวเลข
df["Number"] = pd.to_numeric(df["Number"])

print(df.head(20))
print("จำนวนค่าว่าง =", df["Number"].isnull().sum())
```

```python

# แปลงวันที่และจำนวนให้เป็นชนิดข้อมูลที่เหมาะสม
df["date"] = pd.to_datetime(df["date"], errors="coerce")

df["Number"] = (
    df["Number"]
    .astype(str)
    .str.strip()
    .str.replace(",", "", regex=False)
    .replace("-", pd.NA)
)

df["Number"] = pd.to_numeric(df["Number"], errors="coerce")

# แสดงผลตรวจสอบ
print("===== ชนิดข้อมูล =====")
print(df.dtypes)

print("\n===== จำนวนค่าว่าง =====")
print(df.isna().sum().to_string())

print("\n===== จำนวนแถวซ้ำ =====")
print(df.duplicated().sum())

print("\n===== ช่วงวันที่ =====")
print("วันแรก:", df["date"].min())
print("วันสุดท้าย:", df["date"].max())
print("จำนวนวันที่ไม่ซ้ำ:", df["date"].nunique())

print("\n===== ตัวอย่างวันที่ไม่ซ้ำ 15 ค่าแรก =====")
print(df["date"].drop_duplicates().sort_values().head(15).to_list())

print("\n===== ขนาดข้อมูลหลังตรวจสอบ =====")
print(df.shape) 
```

```python

# เก็บปีจากวันที่เดิม
df["Year"] = df["date"].dt.year

# ใช้เลขวันเดิมแทนเลขเดือน
df["Month"] = df["date"].dt.day

# สร้างวันที่ใหม่ให้เป็นวันแรกของเดือนที่ถูกต้อง
df["date"] = pd.to_datetime({
    "year": df["Year"],
    "month": df["Month"],
    "day": 1
})

# ตรวจสอบผล
print("===== วันที่หลังแก้ไข =====")
print(df["date"].drop_duplicates().sort_values().to_list())

print("\n===== จำนวนเดือนในแต่ละปี =====")
print(df.groupby("Year")["Month"].nunique().to_string())

print("\n===== ช่วงวันที่ใหม่ =====")
print("วันแรก:", df["date"].min())
print("วันสุดท้าย:", df["date"].max())
print("จำนวนเดือนทั้งหมด:", df["date"].nunique())

print("\n===== ขนาดข้อมูล =====")
print(df.shape)
```

```python

# สร้างตัวแปรปี เดือน และไตรมาส
df["Year"] = df["date"].dt.year
df["Month"] = df["date"].dt.month
df["Quarter"] = df["date"].dt.quarter

# แสดงตัวอย่างข้อมูล
print(df[["date", "Year", "Month", "Quarter",
          "Country", "Number"]].head(10).to_string(index=False))

# ตรวจสอบจำนวนข้อมูล
print("\nจำนวนแถวและคอลัมน์:", df.shape)
```

```python

# นับจำนวนรายการและจำนวนข้อมูลที่มีตัวเลขในแต่ละปี
check = df.groupby("Year")["Number"].agg(
    ["size", "count"]
)

# เปลี่ยนชื่อคอลัมน์ให้อ่านง่าย
check.columns = ["จำนวนรายการ", "จำนวนข้อมูลที่มี"]

# คำนวณจำนวนข้อมูลที่หายไป
check["จำนวนข้อมูลที่หายไป"] = (
    check["จำนวนรายการ"] - check["จำนวนข้อมูลที่มี"]
)

# แสดงผล
print(check.to_string())
```

```python

# ตรวจสอบข้อมูลที่หายไปในแต่ละไตรมาสของแต่ละปี
check_quarter = df.groupby(["Year", "Quarter"])["Number"].agg(
    ["size", "count"]
)

check_quarter.columns = ["จำนวนรายการ", "จำนวนข้อมูลที่มี"]

check_quarter["จำนวนข้อมูลที่หายไป"] = (
    check_quarter["จำนวนรายการ"]
    - check_quarter["จำนวนข้อมูลที่มี"]
)

print(check_quarter.to_string())
```

```python
# ตรวจสอบไตรมาสที่ไม่มีข้อมูล Number เลย
missing_quarters = check_quarter[
    check_quarter["จำนวนข้อมูลที่มี"] == 0
]

print("===== ไตรมาสที่ไม่มีข้อมูล Number =====")
print(missing_quarters.to_string())
```

```python
# ตรวจสอบก่อนว่าแต่ละไตรมาสมีข้อมูลหายไปกี่เปอร์เซ็นต์
check = df.groupby(["Year", "Quarter"])["Number"].agg(
    ทั้งหมด="size",
    มีข้อมูล="count"
)

check["หายไป (%)"] = (
    (check["ทั้งหมด"] - check["มีข้อมูล"])
    / check["ทั้งหมด"] * 100
).round(2)

print(check.to_string())
```

```python
import matplotlib.pyplot as plt

sizes = [165781845, 42767635, 9291281, 3543405, 1141468, 64764]
labels = [
    'Asia & Pacific','Europe','America','Middle East','Africa','Other',
]

plt.figure(figsize=(8, 5))


def show_pct(pct):
  return f'{pct:.1f}%' if pct >= 2 else ''


wedges, texts, autotexts = plt.pie(
    sizes,
    autopct=show_pct,
    pctdistance=0.7,
    startangle=140,
    explode=[0.05, 0, 0, 0, 0, 0],
    wedgeprops=dict(edgecolor='white', linewidth=2),
)

for t in autotexts:
  t.set_color('white')
  t.set_weight('bold')

plt.legend(
    wedges,
    [f'{l} ({s/sum(sizes)*100:.1f}%)' for l, s in zip(labels, sizes)],
    title='Continents',
    loc='center left',
    bbox_to_anchor=(1, 0.5),
)

plt.title('Tourists by Continent')
plt.tight_layout()
plt.show()
```
